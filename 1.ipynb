## 2. Теоретическая справка (кратко)
Любую матрицу $ A $ размером $ m \times n $ можно разложить на три матрицы:
$
A = U \cdot \Sigma \cdot V^T
$
Где:
- $ U $ — ортогональная матрица размера $ m \times m $ (левые сингулярные векторы).
- $ \Sigma $ — диагональная матрица размера $ m \times n $, на диагонали которой стоят сингулярные числа $ \sigma_1 \ge \sigma_2 \ge ... \ge \sigma_k $.
- $ V^T $ — ортогональная матрица размера $ n \times n $ (правые сингулярные векторы).

**Идея сжатия:** Если оставить только первые $ r $ сингулярных чисел (и соответствующие столбцы U и V), мы получим матрицу $ A_r $, которая является наилучшим приближением $ A $ среди всех матриц ранга $ r $ (в смысле нормы Фробениуса).

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image
````

````python
img = matplotlib.image.imread('popugai.jpg') #чтение картинки
R = img[:,:,0] #красный канал
G = img[:,:,1] #зеленый канал
B = img[:,:,2] #синий канал
img_gray = 0.299*R + 0.587*G + 0.114*B #серый
h, w = img_gray.shape
original_flat = img_gray.flatten()

plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.show()
print(h,w)
````

**Часть А:** Превращаем исходную картинку в удобный рабочий материал. Изначальное изображение состоит из пикселей, с которыми в дальнейшем мы не сможем работать, потому преобразуем все в массив. Изображение цветное в 3 канала rgb, а значит, больше чисел, больше данных. Чтобы сократить это, мы переводим в оттенки серого по формуле, и получаем одну полноценную матрицу

````python
U, s, Vt = np.linalg.svd(img_gray, full_matrices=False) #svd
list_of_ranks = [1, 5, 10, 30, 100, min(img_gray.shape)]
original_shape = (h,w)

fig, axes = plt.subplots(2, 3, figsize=(12, 8)) #график

approximations = []
for i,r in enumerate(list_of_ranks):
    Ar = (U[:, :r] * s[:r]) @ Vt[:r, :] #восстановление для ранга r

    approximations.append(Ar)
    axes.flat[i].imshow(Ar, cmap='gray')
    axes.flat[i].set_title(f'Ранг = {r}\nСжатие в {((3*w*h))/(r*(h+1+w)*3*4)} раз')
    axes.flat[i].axis('off')

plt.tight_layout()
plt.show()
````

**Часть Б:** Раскладываем на 3 матрицы и от каждой берем r чисел. Чем меньше ранг, тем картинка менее различима. При ранге r=30 картинка приобретает примерные очертания. При ранге r=100 мы без усилий видим объект, но с шумом. При ранге r=1080 картинка становится хорошего качества без потерь

**Часть В:** Для удобства сравнения выводим все картинки в виде графика 2х3

**Часть Г:** Рассчитываем разницу между исходным изображением и сжатым. Чем меньше ранг, тем больше коэффициент сжатия, и тем картинка меньше весит

````python
# ---------- БЛОК АВТОПРОВЕРКИ (НЕ РЕДАКТИРОВАТЬ) ----------
# Предполагается, что у вас есть переменные:
# original_shape = (h, w, c) или (h, w)
# list_of_ranks = [1, 5, 10, 30, 100, ...]
# approximations = список восстановленных массивов (numpy) для каждого ранга

# 1. Проверка размерностей
for i, r in enumerate(list_of_ranks):
    assert approximations[i].shape == original_shape, f"Ошибка: размерность для ранга {r} не совпадает с исходной"

# 2. Проверка, что значения не выходят за пределы 0-255 (для RGB/серого)
for i, r in enumerate(list_of_ranks):
    arr = approximations[i]
    if arr.dtype != np.uint8:
        arr = np.clip(arr, 0, 255) # Если float, то проверяем диапазон
    assert arr.max() <= 255.1, f"Значения превышают 255 для ранга {r}"
    assert arr.min() >= -0.1, f"Значения меньше 0 для ранга {r}"

# 3. Проверка, что с ростом ранга ошибка уменьшается (метрика MSE)
from sklearn.metrics import mean_squared_error
mse_list = []
for i in range(len(approximations)):
    # Если цветная, считаем MSE по всем каналам
    mse = mean_squared_error(original_flat, approximations[i].flatten())
    mse_list.append(mse)

# Проверяем монотонность (MSE должен падать, так как мы добавляем сингулярные числа)
for i in range(1, len(mse_list)):
    assert mse_list[i] <= mse_list[i-1] + 1e-6, f"MSE не уменьшается между рангами {list_of_ranks[i-1]} и {list_of_ranks[i]}"

print("✅ Все автоматические проверки пройдены. Задание выполнено корректно!")
````

**Выводы:** Выполнена задача - сжатие растрового изображения с контролируемой потерей качества. Попробовав разные ранги, визуальное качество перестает отличаться от оригинала примерно при r=100. Чем меньше ранг, тем больше сжатие. Чтобы достигнуть сжатия в 10 раз, ранг должен быть приблизительно r=17. В нашем случае изображение при таком сжатии различимо, но с сильной потерей качества. На темных и светлых участках ошибка воспринимается по-разному. Это связано с тем, что у темных пикселей небольшие значения, и ошибка составляет большой процент от их яркости, на светлых участках те же значения составляют не такую большую долю и потому почти незаметны.
